CALCULATING MEAN

METHOD 1 :- MANUAL CALCULATION

```python
population =[150,160,170,165,155]
sample = [150,170,165]

pop_mean = sum(population) / len(population)
sample_mean = sum(sample) / len(sample)

print("Population Mean:", pop_mean)
print("Sample Mean:", sample_mean)
```

METHOD 2 :- USING NUMPY


```python
import numpy as np  
data = [150, 160, 170, 165, 155]

airthmetic_mean = np.mean(data)
print("Arithmetic Mean:", airthmetic_mean)
```

METHOD 3 :- USING STATISTICS

```python
#pip install statistics

import statistics as stats

var =[10,20,30,40,50]

mean_stats = stats.mean(var)
print("Mean using statistics module:", mean_stats)
```

VARIANCE 

USING NUMPY

```python
import numpy as np 

data = [2,4,4,5,6,7,8,9]

sample_variance = np.var(data, ddof=1) # sample variance ddof "delata degrees of freedom"
pop_variance = np.var(data,ddof=0) # population variance
print("Sample Variance:", sample_variance)
print("Population Variance:", pop_variance)

```

USING PANDAS 

```python
import pandas as pd

data = [2,5,6,67,8,5,4,9,14,2,10]
series = pd.Series(data)    

sample_variance = series.var(ddof=1) # sample variance ddof "delata degrees of freedom"
print("Sample Variance using pandas:", sample_variance)

pop_variance = series.var(ddof=0) # population variance
print("Population Variance using pandas:", pop_variance)
```

USING SCIPY

```python
from scipy import stats 

data = [7,8,9,1,8,5,2,6,4,7]
sample_variance = stats.tvar(data) # sample variance
print("Sample Variance using scipy:", sample_variance)

pop_variance = stats.tvar(data, ddof=0)  # Correctly calculates population variance
print("Population Variance using scipy:", pop_variance)
```

DATA REPRESENTATION


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#CREATING A FIGURE AND AXES

fig,axs = plt.subplots(3,2, figsize=(14,12))
fig.suptitle('Histogram With KDE of Various Distribution',fontsize=16)

#normal distribution (gaussian distribution)
data = np.random.normal(loc=0, scale=1, size=1000) #mean=0, std Dev=1
sns.histplot(data, kde=True, ax=axs[0,0], color='blue',stat='density',bins=30)
axs[0,0].set_title('Normal Distribution')
axs[0,0].grid(True)

#skewed distribution
data_right_skewed = np.random.exponential(scale=1, size=1000) #mean=1, std Dev=1
data_left_skewed = np.random.exponential(scale=1, size=1000)

#Right skewed histrogram
sns.histplot(data_right_skewed, kde=True, ax=axs[0,1], color='Red',stat='density',bins=30,label ='Right Skewed')

#left skewed histogram
sns.histplot(data_left_skewed, kde=True, ax=axs[0,1], color='Green',stat='density',bins=30,label ='Left Skewed')
axs[0,1].set_title('Skewed Distribution')
axs[0,1].legend()
axs[0,1].grid(True)

#uniform distribution
data_uniform = np.random.uniform(low=0, high=1, size=1000) #mean=0.5, std Dev=0.29
sns.histplot(data_uniform, kde=True, ax=axs[1,0], color='purple',stat='density',bins=30)
axs[1,0].set_title('Uniform Distribution')
axs[1,0].grid(True)

#bimodal distribution
data_bimodal = np.concatenate([np.random.normal(loc=-2, scale=1, size=500), np.random.normal(loc=3, scale=1, size=500)]) #mean=-2, std Dev=1
sns.histplot(data_bimodal, kde=True, ax=axs[1,1], color='orange',stat='density',bins=30)
axs[1,1].set_title('Bimodal Distribution')
axs[1,1].grid(True)  

#multimodal distribution

data_multimodal = np.concatenate([np.random.normal(loc=-2, scale=1, size=500),
                                 np.random.normal(loc=0, scale=1, size=500), 
                                 np.random.normal(loc=3, scale=1, size=500)]) #mean=-2, std Dev=1   
sns.histplot(data_multimodal, kde=True, ax=axs[2,0], color='pink',stat='density',bins=30)
axs[2,0].set_title('Multimodal Distribution')
axs[2,0].grid(True)

#hide empty subplot
axs[2,1].axis('off')

#adjust layout
plt.tight_layout(rect=[0, 0.0, 1, 0.95])
plt.show()

```
