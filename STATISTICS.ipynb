# VARIANCE 

# USING NUMPY

```python
import numpy as np 

data = [2,4,4,5,6,7,8,9]

sample_variance = np.var(data, ddof=1) # sample variance ddof "delata degrees of freedom"
pop_variance = np.var(data,ddof=0) # population variance
print("Sample Variance:", sample_variance)
print("Population Variance:", pop_variance)

```

# USING PANDAS 

```python
import pandas as pd

data = [2,5,6,67,8,5,4,9,14,2,10]
series = pd.Series(data)    

sample_variance = series.var(ddof=1) # sample variance ddof "delata degrees of freedom"
print("Sample Variance using pandas:", sample_variance)

pop_variance = series.var(ddof=0) # population variance
print("Population Variance using pandas:", pop_variance)
```

# T TEST AND P TEST

```python
import numpy as np  
import scipy.stats as stats

#step 1:- Generate two random samples from normal distributions
# Generate two random samples from normal distributions
group_A = np.random.normal(75,10,30)
group_B = np.random.normal(80,12,30)

#step 2:-perform an independent t-test
# Generate two random samples from normal distributions
t_statistic, p_value = stats.ttest_ind(group_A, group_B)
#step3:- Displayes the t-statistic and p-value
print("T-statistic:", t_statistic)
print("P-value:", p_value)
# step4 :- interpretation based on p-value
# Null Hypothesis (H0): There is no significant difference between the two groups.
alpha = 0.05 # significance level
if p_value <= alpha:
    print("Reject the null hypothesis: There is a significant difference between the two groups.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the two groups.") 
```

# Z TEST

IMAGINE A COMPANY CLAIMS THAT THE AVG WEIGHT OF THEIR PRODUCT IS 500 GM . WE WANT TO TEST IF THIS CLAIMS IS TRUE BY COLLECTING A SAMPLE OF 30 PRODUCTS  THE SAMPLE MEAN IS 505 GM , AND THE POPULATION STANDARD DEVIATION is 10 gram  .
We will conduct our Z-test to see if the difference in the mean is significant or not

```python
import numpy as np
import scipy.stats as stats 

#step1 :- define the known parameters of the population
population_mean = 500 # claimed population mean(mu)
sample_mean = 500 # sample mean(x bar)
population_std = 10 # population standard deviation(sigma)
sample_size = 30 # sample size(n)

#step 2:- calculate the z-Value

z_value = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))
print("Z-Value:", z_value)

#step 3:- calculate the p-value(two_tailed test)

p_value = 2 * (1 - stats.norm.cdf(abs(z_value)))
print("P-Value:", p_value)

#step:=4:- set the significance level 

alpha = 0.05

#step 5 :- compare p-value with significant level
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference between the sample mean and population mean.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the sample mean and population mean.")
```
